```python
import requests

# 1. Configuration
# BASE_URL = "https://re-api.enersoln.org/weather/pull/vayu"
BASE_URL = "http://localhost:5000/weather/pull/vayu"
API_KEY = "rnsghtT@89)"

# 2. Query Parameters
params = {
    "model_name": "ncm_d2",
    "start_date": "2026-03-27T00:00:00",
    "end_date": "2026-03-29T15:00:00"
}

# 3. Request Headers
headers = {
    "X-API-Key": API_KEY,
}

print(f"Sending request to: {BASE_URL}...")

try:
    # Send GET request
    response = requests.get(BASE_URL, headers=headers, params=params, timeout=10)

    # Print Response Info
    print(f"Status Code: {response.status_code}")
    print("--- Response Body ---")
    
    # Try parsing JSON output
    try:
        print(response.json())
    except ValueError:
        # If the response isn't JSON (e.g. raw text or HTML)
        print(response.text)

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```
